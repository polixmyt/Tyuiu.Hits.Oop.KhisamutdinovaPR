<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic; //Работа с коллекциями

// Базовый класс Task
public class Task
{
    public int TaskId { get; set; } //Номер
    public string TaskName { get; set; } // Название
    public string Priority { get; set; } // Приоритет
    public bool IsCompleted { get; protected set; } // Правда, Ложь
    public string AssignedTo { get; protected set; } // Имя

    public Task(int taskId, string taskName, string priority, string assignedTo = "Не назначен")
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = assignedTo;
        IsCompleted = false;
    }

    public virtual void MarkAsComplete() // метод, который ничего не возвращает
    {
        IsCompleted = true;
        Console.WriteLine($"[Задача #{TaskId}] '{TaskName}' отмечена как выполненная.");
    }

    public virtual string GetTaskDetails()
    {
        string status = IsCompleted ? "Завершено" : "В работе";
        return $"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | Исполнитель: {AssignedTo} | Статус: {status}";
    }

    public virtual void ReassignTo(string newAssignee) // метод для смены исполнителя
    {
        AssignedTo = newAssignee;
        Console.WriteLine($"[Задача #{TaskId}] Переназначена на исполнителя: {AssignedTo}");
    }
}

// 1. Производный класс 
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; } // дедлайн

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate, string assignedTo = "Не назначен")
        : base(taskId, taskName, priority, assignedTo)
    {
        DueDate = dueDate;
    }
    public class Math 

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[Делегатская задача #{TaskId}] '{TaskName}' выполнена к сроку: {DueDate:dd.MM.yyyy}.");
    }
}

// 2. Производный класс 
public class TeamTask : Task // Задача у конкретной группы
{
    public string TeamName { get; set; }

    public TeamTask(int taskId, string taskName, string priority, string teamName, string assignedTo = "Не назначен")
        : base(taskId, taskName, priority, assignedTo)
    {
        TeamName = teamName;
    }

    public override void ReassignTo(string newMember)
    {
        AssignedTo = newMember;
        Console.WriteLine($"[Командная задача #{TaskId}] В команде '{TeamName}' задача переназначена на участника: {AssignedTo}");
    }
}

// 3. Производный класс ResearchTask
public class ResearchTask : Task //добавляет ифнормацию  об источнике данных
{
    public string DataSource { get; set; }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource, string assignedTo = "Не назначен")
        : base(taskId, taskName, priority, assignedTo)
    {
        DataSource = dataSource;
    }

    public override string GetTaskDetails()
    {
        return $"{base.GetTaskDetails()} | Источник данных: {DataSource}";
    }
}

// Демонстрация работы
List<Task> projectTasks = new List<Task>
{
    new DelegateTask(101, "Подготовить отчет", "Высокий", DateTime.Now.AddDays(3), "Артем"),
    new TeamTask(102, "Разработка модуля", "Средний", "AlphaTeam", "Александр"),
    new ResearchTask(103, "Анализ рынка", "Низкий", "База данных Kaggle", "Роман")
};

Console.WriteLine("=== СПИСОК ВСЕХ ЗАДАЧ ===");
foreach (var task in projectTasks)
{
    Console.WriteLine(task.GetTaskDetails());
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");
projectTasks[0].MarkAsComplete();
projectTasks[1].ReassignTo("Альбина");
Console.WriteLine(projectTasks[2].GetTaskDetails());
```
