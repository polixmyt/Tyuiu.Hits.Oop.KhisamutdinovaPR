Система управления задачами (Task Management System)

Цель работы — спроектировать и реализовать архитектуру классов с демонстрацией всех видов наследования:
1. Простое наследование: базовый класс Task и производные классы DelegateTask и ResearchTask.
2. Сложное (многоуровневое) наследование: цепочка наследования Task → TeamTask → UrgentTeamTask с расширением логики через ключевое слово base.
3. Множественное наследование: реализация контрактов интерфейсов INotifiable и IReportable для отправки уведомлений и формирования отчетов.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;


public interface INotifiable
{
    void SendNotification(string message);
}

public interface IReportable
{
    void GenerateReport();
}


public class Task
{
    public string TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public bool IsCompleted { get; set; }

    public Task(string taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        IsCompleted = false;
    }

    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Задача '{TaskName}' отмечена как выполненная.");
    }

    public virtual void GetTaskDetails()
    {
        string status = IsCompleted ? "Выполнена" : "В процессе";
        Console.WriteLine($"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | Статус: {status}");
    }

    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача '{TaskName}' переназначена на: {newAssignee}.");
    }

    public void UpdatePriority(string newPriority)
    {
        Priority = newPriority;
        Console.WriteLine($"Приоритет задачи '{TaskName}' изменен на: {Priority}.");
    }
}

public class DelegateTask : Task
{
    public string DueDate { get; set; }

    public DelegateTask(string taskId, string taskName, string priority, string dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }


    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Делегированная задача '{TaskName}' завершена к сроку: {DueDate}.");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"  Срок выполнения: {DueDate}");
    }
}


public class TeamTask : Task, INotifiable
{
    public string TeamName { get; set; }

    public TeamTask(string taskId, string taskName, string priority, string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }


    public override void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача '{TaskName}' переназначена участнику команды '{TeamName}': {newAssignee}.");
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"  Команда: {TeamName}");
    }


    public void SendNotification(string message)
    {
        Console.WriteLine($"[Уведомление для команды '{TeamName}']: {message}");
    }
}


public class ResearchTask : Task
{
    public string DataSource { get; set; }

    public ResearchTask(string taskId, string taskName, string priority, string dataSource)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    
    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"  Источники данных: {DataSource}");
    }
}


public class UrgentTeamTask : TeamTask, IReportable
{
    public string UrgencyReason { get; set; }

    public UrgentTeamTask(string taskId, string taskName, string priority, string teamName, string urgencyReason)
        : base(taskId, taskName, priority, teamName)
    {
        UrgencyReason = urgencyReason;
    }

    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"  ВНИМАНИЕ! Задача срочная, причина: {UrgencyReason}");
    }


    public void GenerateReport()
    {
        Console.WriteLine($"[СРОЧНЫЙ ОТЧЕТ] Задача #{TaskId} '{TaskName}' команды '{TeamName}'. Причина срочности: {UrgencyReason}.");
    }
}

Console.WriteLine("--- Тест DelegateTask ---");
DelegateTask delTask = new DelegateTask("1", "Сдать лабораторную", "Высокий", "05.10.2026");
delTask.GetTaskDetails();
delTask.MarkAsComplete();

Console.WriteLine("\n--- Тест ResearchTask ---");
ResearchTask resTask = new ResearchTask("2", "Изучить фреймворк", "Средний", "Документация Microsoft, Habr");
resTask.GetTaskDetails();

Console.WriteLine("\n--- Тест TeamTask ---");
TeamTask teamTask = new TeamTask("3", "Разработка модуля авторизации", "Высокий", "Backend");
teamTask.ReassignTo("Иван");
teamTask.SendNotification("Собрание по задаче в 15:00.");

Console.WriteLine("\n--- Тест UrgentTeamTask ---");
UrgentTeamTask urgentTask = new UrgentTeamTask("4", "Исправление критического бага", "Критический", "DevOps", "Упал продакшн сервер");
urgentTask.ReassignTo("Анна");
urgentTask.SendNotification("Сервер не отвечает!");
urgentTask.GenerateReport();
```
